```python
from spellchecker import SpellChecker

# Create spell checker
spell = SpellChecker()

# Get text from user
text = input("Enter a sentence: ")

# Split words
words = text.split()

print("\nSPELL CHECKING")
print("--------------------")

corrected_words = []

for word in words:

    # Remove punctuation
    clean_word = word.strip(".,!?;:")

    # Check spelling
    if clean_word.lower() in spell:
        print(clean_word, "✓ Correct")
        corrected_words.append(clean_word)

    else:
        correction = spell.correction(clean_word)

        print(clean_word, "✗ Incorrect")
        print("Suggestion:", correction)

        corrected_words.append(correction)

# Corrected sentence
corrected_sentence = " ".join(corrected_words)

print("\nCorrected Sentence:")
print(corrected_sentence)
```
